**ELEIÇÃO**

```python
# Crie uma lista de candidatos, identificados por números.
# Crie uma estrutura para armazenar a quantidade de votos de cada candidato.
# Cadastre os eleitores, solicitando nome e idade.
# Verifique a situação eleitoral de acordo com a idade:
# Menor de 16 anos: não pode votar;
# 16 e 17 anos: voto opcional;
# 18 anos ou mais: voto obrigatório.
# Para os eleitores aptos a votar, apresente os candidatos e permita a escolha.
# Valide o número do candidato informado.
# Contabilize os votos e a quantidade de eleitores que participaram da votação.
# Permita o cadastro de vários eleitores até que o usuário escolha encerrar.
# Ao finalizar, apresente:
# total de eleitores cadastrados;
# total de eleitores que votaram;
# quantidade de votos opcionais;
# quantidade de votos obrigatórios;
# votos de cada candidato.
# Identifique o candidato vencedor.
# Caso dois ou mais candidatos tenham a mesma quantidade de votos, o sistema deverá informar que houve empate.
# Realize testes com diferentes idades, candidatos e situações de votação.

# ==========================================
# SISTEMA DE VOTAÇÃO
# ==========================================

# Lista de candidatos
candidatos = {
    1: "Breno Fofinho da Etec",
    2: "Louis aleijadinho",
    3: "GAYbriel",
    4: "Marina Menina Linda"
}

# Estrutura para armazenar os votos de cada candidato
votos = {
    1: 0,
    2: 0,
    3: 0,
    4: 0
}

# Contadores
total_eleitores = 0
eleitores_que_votaram = 0
votos_opcionais = 0
votos_obrigatorios = 0



# ==========================================
# CADASTRO DOS ELEITORES
# ==========================================

while True:

    print("\n===== CADASTRO DE ELEITOR =====")

    nome = input("Digite o nome do eleitor: ")
    idade = int(input("Digite a idade do eleitor: "))

    total_eleitores += 1

    # ======================================
    # VERIFICAÇÃO DA SITUAÇÃO ELEITORAL
    # ======================================

    if idade < 16:
        print(f"\n{nome}, você não pode votar.")
        print("Idade inferior a 16 anos.")

    elif idade == 16 or idade == 17:
        print(f"\n{nome}, seu voto é opcional.")

        votar = input("Deseja votar? (S/N): ").upper()

        if votar == "S":

            print("\n===== CANDIDATOS =====")

            for numero, nome_candidato in candidatos.items():
                print(f"{numero} - {nome_candidato}")

            # Validação do candidato
            while True:
                try:
                    numero_voto = int(input("Digite o número do candidato: "))

                    if numero_voto in candidatos:
                        break
                    else:
                        print("Número de candidato inválido. Tente novamente.")

                except ValueError:
                    print("Digite apenas um número.")

            votos[numero_voto] += 1
            eleitores_que_votaram += 1
            votos_opcionais += 1

            print("Voto registrado com sucesso!")

        else:
            print("Eleitor optou por não votar.")

    else:
        print(f"\n{nome}, seu voto é obrigatório.")

        print("\n===== CANDIDATOS =====")

        for numero, nome_candidato in candidatos.items():
            print(f"{numero} - {nome_candidato}")

        # Validação do candidato
        while True:
            try:
                numero_voto = int(input("Digite o número do candidato: "))

                if numero_voto in candidatos:
                    break
                else:
                    print("Número de candidato inválido. Tente novamente.")

            except ValueError:
                print("Digite apenas um número.")

        votos[numero_voto] += 1
        eleitores_que_votaram += 1
        votos_obrigatorios += 1

        print("Voto registrado com sucesso!")

    # ======================================
    # PERGUNTA SE DESEJA CONTINUAR
    # ======================================

    continuar = input("\nDeseja cadastrar outro eleitor? (S/N): ").upper()

    if continuar != "S":
        break


# ==========================================
# RESULTADO FINAL
# ==========================================

print("\n")
print("==========================================")
print("           RESULTADO DA VOTAÇÃO")
print("==========================================")

print(f"Total de eleitores cadastrados: {total_eleitores}")
print(f"Total de eleitores que votaram: {eleitores_que_votaram}")
print(f"Quantidade de votos opcionais: {votos_opcionais}")
print(f"Quantidade de votos obrigatórios: {votos_obrigatorios}")

print("\n===== VOTOS POR CANDIDATO =====")

for numero, nome_candidato in candidatos.items():
    print(f"{numero} - {nome_candidato}: {votos[numero]} voto(s)")


# ==========================================
# IDENTIFICAÇÃO DO VENCEDOR
# ==========================================

maior_numero_votos = max(votos.values())

# Verifica quais candidatos possuem a maior quantidade
vencedores = []

for numero, quantidade in votos.items():
    if quantidade == maior_numero_votos:
        vencedores.append(numero)


print("\n===== RESULTADO FINAL =====")

if len(vencedores) == 1:
    vencedor = vencedores[0]

    print(
        f"Vencedor: {candidatos[vencedor]} "
        f"({maior_numero_votos} voto(s))"
    )

else:
    print("Houve empate entre os seguintes candidatos:")

    for numero in vencedores:
        print(
            f"{numero} - {candidatos[numero]} "
            f"({maior_numero_votos} voto(s))"
        )

```
