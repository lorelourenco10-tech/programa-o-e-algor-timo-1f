```python
# Escrever um programa em que o usuário digita um número e o programa faz a contagem regressiva

numero = int(input("Digite um número para iniciar a contagem regressiva: "))

# Começa em um número, vai até -1 (para incluir o 0) e subtrai 1 a cada volta
for i in range (numero, -1, -1):
  print (i)
```

```python
# Escrever um programa em que o usúario digita um número e o programa soma os números até o número digitado

numero = int (input("Digite um número: "))
soma = 0

for i in range (1, numero, +1):
  soma += i
print (f"A soma dos números de 1 até {numero} é: {soma}.")
```

```python
# Escrever um programa em que o usuário digita uma palavra e o programa conta as vogais da palavra
palavra = str (input("Digite uma palavra: "))
vogais = "a, e, i, o, u, A, E, I, O, U"

for letra in palavra:
  if letra in vogais:
    print (f"{letra}")
```
